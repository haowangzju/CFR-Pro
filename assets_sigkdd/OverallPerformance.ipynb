```python
import os
import numpy as np
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
from matplotlib.backends.backend_pdf import PdfPages
import seaborn as sns
matplotlib.rcParams['pdf.fonttype'] = 42
matplotlib.rcParams['ps.fonttype'] = 42
plt.rc('font', family='Arial')
matplotlib.rcParams['mathtext.fontset'] = 'stix'
matplotlib.rcParams['font.size'] = 10
pd.set_option('display.max_rows', 1000)
root = '.'
```

```python
def load_yaml_file(file_path):
    import yaml
    try:
        with open(file_path, encoding='utf-8') as file:
            return yaml.load(file.read(), Loader=yaml.FullLoader)
    except FileNotFoundError as e:
        print(f"Error: {e}")
        return None


def load_data(roots=['./baseline0518/ACIC/descn', './baseline0518/ACIC/site'], exist_config=True):

    assert type(roots) is list
    data = []

    for root in roots:
        dirs = os.listdir(f"{root}")
        for dir in dirs:
            if '.md' in dir:
                continue
            _data = pd.read_csv(f"{root}/{dir}/run.txt",
                                sep=",", header=0, index_col=None)
            if exist_config:
                _configs = load_yaml_file(f"{root}/{dir}/config.yaml")
                for key in _configs:
                    _data[key] = _configs[key]
            else:
                _data['model'] = dir.split('_')[0]
            data.append(_data)
    return pd.concat(data)
```

## Out-of-sample ACIC

```python
data = load_data(['./baseline_final/ACIC'], exist_config=False)
data = data.query("model != 'CForest'")
data.query("mode=='test'").groupby('model').mean(numeric_only=True).round(4).astype(str) + '±' + data.query("mode=='test'").groupby('model').std(numeric_only=True).round(4).astype(str)
```
